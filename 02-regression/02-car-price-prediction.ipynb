```python
import matplotlib.pyplot as plt
import seaborn as sns

%matplotlib inline
```

```python
np.log( [ 1 , 10 , 1000 , 100000])
# if we write 0 -> np.log(0) we will get -inf, so we need to filter out the 0s before taking the log.
#runtimeWarning: divide by zero encountered in log
```

```python
np.log( [0+1, 1 +1, 10 +1, 1000 +1 , 100000 +1])
# so we can add 1 everywhere to avoid the -inf, but we need to be careful because it will change the distribution of the data. So we need to make sure that we are not changing the distribution of the data too much.
```

```python
# we can use this function in numpy too : log1p , to avoid the -inf, and it will not change the distribution of the data too much.
np.log1p([0, 1, 10, 1000, 100000])
```

```python
price_logs = np.log1p(df.msrp)
price_logs
```

```python
sns.histplot( price_logs , bins = 50)
# there is no tail anymore and the distribution is more normal, so we can use this transformation to make the distribution more normal and then we can use linear regression on it.
```
